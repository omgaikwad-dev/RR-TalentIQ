# Notebook 01: IPL Data Exploration (2008–2026)

This notebook demonstrates loading, inspecting, and validating the raw and cleaned Cricsheet datasets.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

matches_df = pd.read_csv('../data/processed/matches_cleaned.csv')
deliveries_df = pd.read_parquet('../data/processed/deliveries_cleaned.parquet')

print(f"Matches shape: {matches_df.shape}")
print(f"Deliveries shape: {deliveries_df.shape}")
matches_df.head(3)
```

## Match Distribution by Season

```
season_counts = matches_df['season'].value_counts().sort_index()
plt.figure(figsize=(10, 4))
season_counts.plot(kind='bar', color='#2b5c8f')
plt.title('Matches Played per Season (2008-2026)')
plt.xlabel('Season')
plt.ylabel('Matches')
plt.show()
```

## Delivery Overview & Boundaries

```
print("Total Fours:", deliveries_df['is_boundary_four'].sum())
print("Total Sixes:", deliveries_df['is_boundary_six'].sum())
print("Total Wickets:", deliveries_df['is_wicket'].sum())
```
